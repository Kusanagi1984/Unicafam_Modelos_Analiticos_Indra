**Taller Big Data**

```python
# Instalación de bibliotecas

import pandas as pd
import numpy as np
```

```python
#se números aleatorios
np.random.seed(42)
n = 200
```

```python
#Generamos nuestra tabla de ventas.
ventas = pd.DataFrame({
    "fecha": pd.date_range(start="2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellin", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n)
})
```

```python
#creación de la columna total de ventas.
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
#imprimir el tamaño de la tabla de ventas
print(ventas.shape)
```

```python
#Imprimir las primeras 10 filas de la tabla
print(ventas.head(10))
```

```python
# Conocer los datos y tipos de datos
ventas.info()
```

```python
# Clasificamos las variables
for col in ventas.columns:
    serie = ventas[col]
    unicos = serie.nunique()
    if pd.api.types.is_datetime64_any_dtype(serie):
        clase = "Fecha"
    elif pd.api.types.is_numeric_dtype(serie):
        clase = "Cuantitativa"
    else:
        clase = "Cualitativa"
    print(f"{col:10} {clase:13} valores únicos: {unicos}")
```

**ANALÍTICA DESCRIPTIVA: ¿QUÉ PASÓ?**


```python
# Hacemos analítica descriptiva - tabla dinamica

resumen = (ventas.groupby("ciudad")["total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))


```

```python
#resumen de los datos
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["ventas", "total_millones", "ticket_prom"]

```

```python
print(resumen)
```
