```python
presupuesto = 4800
precio_cafe = 67
cafes_semana = 4



# 1. ¿Cuánto gastas al mes? (4 semanas)
gasto_mes = precio_cafe * (cafes_semana * 4)

# 2. ¿Cuánto te sobra?
sobra = presupuesto - gasto_mes

# 3. ¿Te alcanza? (True o False)
alcanza = presupuesto >= gasto_mes

# 4. Imprime los resultados
print("gasto mensual: ", gasto_mes, "sobra: ", sobra, "alcanza: ",alcanza)
```

**TEMA 8 · BUCLE FOR**
Problema: revisa tu ticket de compra

```python
# ← cambia por los precios de tu compra
precios = [400, 200, 50, 10]



# 1. Imprime cada precio con un for
for precio in precios:
  print("el valor del producto es ",precio)


# 2. Suma el total
total = 0

for precio in precios:
  total += precio

print("el total de la compra es: ",total)



# 3. ¿Cuántos cuestan más de 100?
caros = 0
for precio in precios:
  if precio > 100:
    caros += 1

print("la cantidad de productos caros es: ",caros)
```

**TEMA 9 · BUCLE WHILE**
Problema: tu meta de ahorro

```python

ahorro = 0
meta = 5000
por_mes = 800
meses = 0



# Mientras no llegues a la meta:

# suma lo del mes y cuenta un mes más

while ahorro < meta:
  ahorro += por_mes
  meses += 1

sobrante = ahorro - meta


# ¿Cuántos meses tardaste?

print(f"el total de meses en llegar a la meta es {meses} y sobra ${sobrante}")
```
