### Clasificación de las Variables

Aquí tienes la clasificación detallada de cada una de las variables del DataFrame:

1. **edad**: **Cuantitativa discreta**. Representa una cantidad contable de años enteros (generalmente se mide en años completos, aunque el tiempo en sí sea continuo).
2. **programa**: **Cualitativa nominal**. Representa categorías (carreras universitarias) sin un orden jerárquico intrínseco.
3. **promedio**: **Cuantitativa continua**. Puede tomar cualquier valor real dentro de un rango (por ejemplo, entre 0.0 y 5.0).
4. **materias_aprobadas**: **Cuantitativa discreta**. Representa un conteo exacto de elementos enteros (no se pueden aprobar 1.5 materias).
5. **jornada**: **Cualitativa nominal**. Categorías de horario de estudio sin un orden de jerarquía.

### **Lienzo de Diseño de Proyecto: CRISP-DM**

#### **Fase 1: Comprensión del Negocio**
*   **Objetivo de Negocio:** Reducir el porcentaje de pedidos que se entregan tarde en Bogotá para mejorar la satisfacción del cliente y la retención de usuarios.
*   **KPI de Éxito:**
    *   Reducir la tasa de entregas que se realizan tarde.

#### **Fase 2: Comprensión de los Datos **
Para resolver este problema, necesitaríamos recopilar y analizar los siguientes 5 datos clave:
1.  **Historial de entregas:** Hora de creación del pedido, hora de salida de la tienda, hora de llegada del repartidor y hora estimada original (ETA).
2.  **Datos geográficos y de tráfico:** Coordenadas de origen (restaurante/tienda), destino (cliente) y estado del tráfico en tiempo real en las localidades de Bogotá.
3.  **Datos del clima:** Registro meteorológico al momento del pedido (lluvia, llovizna, día despejado), dado que la lluvia en Bogotá impacta drásticamente la movilidad.
4.  **Detalles del pedido:** Tipo de comida/producto, volumen, peso y tiempo de preparación estimado por el comercio.
5.  **Información del repartidor:** Tipo de vehículo utilizado (bicicleta, moto, carro), experiencia del repartidor y su ubicación al momento de la asignación.



#### **Fase 3: Preparación de los Datos (Data Preparation)**
Dos problemas de calidad de datos muy comunes que esperaríamos encontrar y deberíamos solucionar son:
1.  **Valores nulos o erróneos en marcas de tiempo:** Registros donde la hora de entrega final es anterior a la hora de creación, o simplemente campos vacíos debido a fallas de conexión GPS en la app del repartidor.
2.  **Ubicaciones GPS atípicas o fuera de rango:** Coordenadas geográficas erráticas (por ejemplo, puntos que se ubican fuera del perímetro de Bogotá o en el océano) causadas por pérdida temporal de señal satelital en los teléfonos de los repartidores o clientes.



#### **Fases 4-6: Modelado, Evaluación e Implantación**
*   **¿Qué predeciríamos? (Modelado):**
    *   Un modelo predictivo para estimar de manera precisa el **Tiempo de Preparación del Pedido** en el restaurante/tienda.
    *   Un modelo de regresión para predecir el **Tiempo de Viaje** óptimo del repartidor considerando el tráfico, clima y tipo de vehículo.
    *   La suma de ambos nos dará un **tiempo de entrega predictivo dinámico y realista** antes de que el cliente realice el pago.
*   **¿Cómo lo usaría la empresa? (Implantación):**
    *   **Asignación Inteligente:** La app asignará el pedido al repartidor que realmente esté más cerca y cuyo medio de transporte sea ideal para la zona y el clima actual.
    *   **Gestión de Expectativas:** Mostrar al cliente un tiempo de entrega más preciso en la pantalla, reduciendo la frustración por promesas incumplidas.

    *   **Alertas Tempranas:** Si el sistema detecta que un pedido en curso superará el límite del retraso permitido, puede enviar una alerta automática al equipo de soporte para priorizar el caso o compensar preventivamente al cliente.

### **Ejercicio 3: Primer análisis de ventas**

Primero, crearemos un DataFrame sintético llamado `ventas` con datos representativos de ciudades de Colombia, canales de venta (Tienda, Web, App) y fechas para realizar nuestro análisis.

```python
import pandas as pd
import numpy as np

# Configurar semilla
np.random.seed(42)

# Crear datos ficticios de ventas
n_registros = 200

ventas = {
    'fecha': pd.date_range(start='2026-10-01', periods=n_registros, freq='D'),
    'canal': np.random.choice(['Tienda', 'Web', 'App'], size=n_registros, p=[0.4, 0.3, 0.3]),
    'ciudad': np.random.choice(['Bogotá', 'Medellín', 'Cali', 'Barranquilla'], size=n_registros),
    'unidades': np.random.randint(1, 20, n_registros),
    "precio": np.random.choice([25000, 48000, 99000], n_registros)
}

#creación de la columna total de ventas.
ventas["total"] = ventas["unidades"] * ventas["precio"]


ventas = pd.DataFrame(ventas)
ventas['fecha'] = pd.to_datetime(ventas['fecha'])
ventas["total"] = ventas["total"].map('${:,.0f}'.format)

# Mostrar las primeras filas del DataFrame
display(ventas.head())
print(f"Total de registros cargados: {len(ventas)}")
```

**PREGUNTAS**

```python
# 1. ¿Cuál es el total vendido por canal (Tienda, Web, App)?
total_por_canal = ventas.groupby('canal')['precio'].sum().reset_index()
total_por_canal.columns = ['Canal de Venta', 'Total Vendido (COP)']
total_por_canal['Total Vendido (COP)'] = total_por_canal['Total Vendido (COP)'].map('${:,.0f}'.format)

print("--- Total Vendido por Canal ---")
display(total_por_canal)
```

```python
# 2. ¿Cuál es el ticket promedio por ciudad?

# Convertimos 'total' de nuevo a numérico eliminando '$' y comas para poder calcular estadísticas
total_numerico = ventas["total"].astype(str).str.replace('$', '').str.replace(',', '').astype(float)

resumen = (ventas.assign(total_num=total_numerico)
                 .groupby("ciudad")["total_num"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))

resumen["sum"] = (resumen["sum" ] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["ventas", "total_millones", "ticket_prom"]
resumen["ticket_prom"] = resumen["ticket_prom"].map("${:,.0f}".format)
print(resumen)
```

```python
# 3. ¿Qué día de la semana se vende más?

# Convertir temporalmente 'total' a tipo numérico
total_numerico = ventas["total"].astype(str).str.replace('$', '').str.replace(',', '').astype(float)
ventas_temp = ventas.assign(total_num=total_numerico)

# Generar tabla dinámica
tabla = pd.pivot_table(ventas_temp, values="total_num", index="ciudad",
                       columns="dia_semana", aggfunc="sum") / 1e6
print("--- Ventas en Millones por Ciudad y Día ---")
print(tabla.round(1))
print()

# Obtener el día de la semana con más ventas totales
dia_mas_vendido = ventas_temp.groupby("dia_semana")["total_num"].sum().idxmax()
print("El día de la semana con más ventas:", dia_mas_vendido)
```
