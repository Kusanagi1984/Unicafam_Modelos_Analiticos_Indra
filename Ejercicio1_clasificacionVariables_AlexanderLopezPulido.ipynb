**Ejercicio 1: Clasifica las variables**

```python
import pandas as pd

# Crear el DataFrame con los datos de 5 estudiantes
data = {
    'edad': [18, 22, 20, 19, 21],
    'estrato': [3, 2, 4, 1, 3],
    'programa': ['Ingeniería', 'Medicina', 'Diseño', 'Derecho', 'Ingeniería'],
    'promedio': [4.2, 3.8, 4.5, 3.9, 4.1],
    'materias_aprobadas': [12, 24, 15, 8, 18],
    'jornada': ['Diurna', 'Nocturna', 'Diurna', 'Diurna', 'Nocturna']
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
display(df)
```

```python
# Revisar los tipos de datos y valores únicos de cada columna
print("--- Tipos de datos ---")
print(df.dtypes)
print("\n--- Cantidad de valores únicos ---")
print(df.nunique())
```

### Clasificación de las Variables

Aquí tienes la clasificación detallada de cada una de las variables del DataFrame:

1. **edad**: **Cuantitativa discreta**. Representa una cantidad contable de años enteros (generalmente se mide en años completos, aunque el tiempo en sí sea continuo).
2. **programa**: **Cualitativa nominal**. Representa categorías (carreras universitarias) sin un orden jerárquico intrínseco.
3. **promedio**: **Cuantitativa continua**. Puede tomar cualquier valor real dentro de un rango (por ejemplo, entre 0.0 y 5.0).
4. **materias_aprobadas**: **Cuantitativa discreta**. Representa un conteo exacto de elementos enteros (no se pueden aprobar 1.5 materias).
5. **jornada**: **Cualitativa nominal**. Categorías de horario de estudio sin un orden de jerarquía.


### Pregunta trampa: ¿El estrato es cuantitativo o cualitativo?

**Respuesta:** El **estrato** es una variable **Cualitativa Ordinal**.

Porque los números establecen un **orden** de jerarquía socioeconómica claramente definido (Estrato 1 < Estrato 2 < Estrato 3...ETC).
