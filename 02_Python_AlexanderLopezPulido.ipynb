**Ejercicio 1: Clasificador de notas**

```python
#funcion para clasificar las notas.
def clasificar_nota(nota):
    if not (0 <= nota <= 5):
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"
```

```python
# Lista de notas a evaluar
import random
notas = [round(random.uniform(1.0, 10.0),2) for _ in range(5)]
```

```python



# Aplicar la función y contar aprobados
aprobados = 0
print("Resultados de la clasificación:")
for nota in notas:
    resultado = clasificar_nota(nota)
    print(f"Nota: {nota} -> {resultado}")
    if resultado in ["Aprueba", "Excelente"]:
        aprobados += 1

print(f"\nCantidad de estudiantes que aprobaron: {aprobados}")
```

**Ejercicio 2: Carrito de compras**

```python
# Definir el carrito de compras: producto -> (precio, cantidad)
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}
```

```python
# Calcular el subtotal
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    subtotal += precio * cantidad
```

```python
# Aplicar descuento si supera los 200.000
descuento = 0
if subtotal > 200000:
    descuento = subtotal * 0.10

subtotal_con_descuento = subtotal - descuento
```

```python
# Calcular el IVA (19%)
iva = subtotal_con_descuento * 0.19

# Total final
total = subtotal_con_descuento + iva
```

```python

# Mostrar resultados con formato de pesos colombianos
print(f"Subtotal original: ${subtotal:,.2f}")
if descuento > 0:
    print(f"Descuento aplicado (10%): -${descuento:,.2f}")
    print(f"Subtotal con descuento: ${subtotal_con_descuento:,.2f}")
print(f"IVA (19%): ${iva:,.2f}")
print(f"Total de la compra: ${total:,.2f}")
```

**Ejercicio 3: Nómina con pandas**

```python
import pandas as pd
import numpy as np

# 1. Crear el DataFrame inicial de empleados
datos_iniciales = {
    'nombre': ['Ana Gómez', 'Carlos Pérez', 'Diana Ruiz'],
    'area': ['Ventas', 'Tecnología', 'Recursos Humanos'],
    'salario': [3500000, 4800000, 3000000]
}
df_empleados = pd.DataFrame(datos_iniciales)
```

```python
# Agregar 3 empleados nuevos usando pd.concat
nuevos_empleados = pd.DataFrame({
    'nombre': ['Luis Martínez', 'Elena Torres', 'Jorge Castro'],
    'area': ['Ventas', 'Administración', 'Tecnología'],
    'salario': [3200000, 2800000, 5200000]
})

df_empleados = pd.concat([df_empleados, nuevos_empleados], ignore_index=True)
```

```python
# 2. Crear la columna bono: 10% del salario si es Ventas, 5% si no
df_empleados['bono'] = np.where(df_empleados['area'] == 'Ventas',
                                df_empleados['salario'] * 0.10,
                                df_empleados['salario'] * 0.05)

# Calcular el salario total (salario + bono)
df_empleados['total_devengado'] = df_empleados['salario'] + df_empleados['bono']

print("--- Tabla de Empleados Actualizada ---")
display(df_empleados)
```

```python
# 3. Calcular el costo total de nómina mensual por área (salario + bono)
nomina_por_area = df_empleados.groupby('area')['total_devengado'].sum().reset_index()
print("\n--- Costo Total de Nómina Mensual por Área ---")
display(nomina_por_area)
```

```python
# 4. ¿Quién es el empleado con mayor salario?
empleado_top = df_empleados.loc[df_empleados['salario'].idxmax()]
print(f"\nEl empleado con mayor salario es {empleado_top['nombre']} con un salario base de ${empleado_top['salario']:,} (Área: {empleado_top['area']})")
```
